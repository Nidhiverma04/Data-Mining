1. Read NIRF data as given in Dataset tab on Google Classroom.

```python
import pandas as pd
```

```python
dataset = pd.read_csv("nirf_engg_ranking_india_2020.csv")
print(dataset.head()) 
```

2. Group the data statewise and Count the number of colleges in each state.

```python
state_group = dataset.groupby("state")
for state, data in state_group:
    print(state,"  :  ",len(data))
```

3.  Sort the records based on rpc attribute (Research Professional practices).

```python
sorted_set = dataset.sort_values(by="rpc", ignore_index=True)
print(sorted_set.head(5))
```

4. Find the rank of Panjab University as per rpc attribute.

```python
print("Rank of Panjab University as per rpc attribute: \n", sorted_set[sorted_set.name == "Panjab University"].index.values[0] + 1)
```

5. List the colleges whose rpc value is less than Panjab University and rank is higher(rank value is less).

```python
rpc_pu = sorted_set[sorted_set.name == "Panjab University"]["rpc"].values[0]
rank_pu = sorted_set[sorted_set.name == "Panjab University"]["rank"].values[0]
data = sorted_set.iloc[26:][sorted_set["rank"] < rank_pu]
print("Total colleges with higher rank and lesser rpc values than PU : ", len(data))
print("\n")
print(data["name"])
```

6. Draw the Histogram for colleges based on tlr values by dividing the overall data in 10 bins.

```python
import matplotlib.pyplot as plt

plt.hist(dataset["tlr"], bins=10, edgecolor="black")
plt.title("Histogram for TLR")
plt.xlabel("tlr values")
plt.ylabel("frequency")
plt.grid(False)
plt.show()
```
