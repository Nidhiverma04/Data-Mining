```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
retail = pd.read_csv("OnlineRetail.csv", encoding="latin1")  
# CSV contains special characters (like £, €, etc.) saved in Windows/Latin encoding, 
# so UTF-8 cannot decode them that's why used latin1
```

1. Find mean, std dev, min , max of unitprice field.

```python
print("mean: ", retail["UnitPrice"].mean())
print("min: ", retail["UnitPrice"].min())
print("max: ", retail["UnitPrice"].max())
print("std: ", retail["UnitPrice"].std())
```

3. Find mean of unitprice group by country

```python
mean_price = retail.groupby("Country")["UnitPrice"].mean()

print(mean_price)
```

4. Find min and max of unit price group by customerID

```python
price_ByCust = retail.groupby("CustomerID")["UnitPrice"]

for id, data in price_ByCust:
    print(f"CustomerID: {id} | Min UnitPrice: {data.min()} | Max UnitPrice: {data.max()}")
```

```python
price_ByCust = (
    retail
    .groupby("CustomerID")["UnitPrice"]
    .agg(["min", "max"])
)

print(price_ByCust)
```

5. Find cumulative sales monthly and yearly.

```python
import pandas as pd
import matplotlib.pyplot as plt

retail["InvoiceDate"] = pd.to_datetime(retail["InvoiceDate"])
retail = retail[retail["UnitPrice"] > 0].copy()
retail["Sales"] = retail["Quantity"] * retail["UnitPrice"]

retail["Month"] = retail["InvoiceDate"].dt.to_period("M")
retail["Year"] = retail["InvoiceDate"].dt.year

monthly_sales = (
    retail
    .groupby("Month")["Sales"]
    .sum()
)

yearly_sales = (
    retail.groupby("Year")["Sales"].sum()
)

cum_monthly = monthly_sales.cumsum()
cum_yearly = yearly_sales.cumsum()

print("Monthly Sales:\n", monthly_sales)
print("\nYearly Sales:\n", yearly_sales)
print("\nCumulative Monthly Sales:\n", cum_monthly)
print("\nCumulative Yearly Sales:\n", cum_yearly)
```
