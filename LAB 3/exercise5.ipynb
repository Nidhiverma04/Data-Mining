```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
import csv

movies = []

with open('Movies.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        movies.append(row)

print(movies[0]['rating'])
```

1.  Compute mean, standard deviation, min, and max of the Rating field.

```python
ratings = {
    'TV-Y'      : 0,    # All children
    'TV-Y7'     : 7,    # Children 7+
    'TV-Y7-FV'  : 7,    # Children 7+ (fantasy violence)
    'TV-G'      : 0,    # General audience
    'G'         : 0,    # General audience
    'TV-PG'     : 10,   # Parental guidance
    'PG'        : 10,   # Parental guidance
    'TV-14'     : 14,   # Teens 14+
    'PG-13'     : 13,   # Teens 13+
    'R'         : 17,   # Restricted
    'TV-MA'     : 17,   # Mature audience
    'NC-17'     : 18,   # Adults only
    'NR'        : None, # Not rated
    'UR'        : None, # Unrated
    ''          : None  # Missing rating
}
```

```python
rating_stats = {}

for movie in movies:
    rating = movie['rating']
    age = ratings.get(rating)

    if age is None:
        continue

    if rating not in rating_stats:
        rating_stats[rating] = {
            'age': age,
            'count': 1
        }
    else:
        rating_stats[rating]['count'] += 1
```

```python
for rating, data in rating_stats.items():
    age = data['age']
    count = data['count']
    mean_rating = age*count

mean_rating /= len(rating_stats)
print(mean_rating)
```
