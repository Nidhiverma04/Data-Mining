```python
def remove_dollar(s):
    if(s.find('$') == 0):
        s = s.replace('$', '')
    return s
```

```python
def remove_comma(s):
    s = s.replace(',', '')
    return int(s)
```

```python
amount = [
    "$40,000",
    "$134",
    "$13,246",
    "$134,578,457",
    "$12"
]

def convert_to_int(s):
    s = remove_dollar(s)
    s = remove_comma(s)
    return s

for (i, entry) in enumerate(amount):
    entry = convert_to_int(entry)
    print(i,': ', entry)
    # print(type(entry))
```
